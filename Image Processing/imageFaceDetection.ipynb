```python
import cv2
```

```python
haar_features = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
```

```python
img = cv2.imread('download.jpg')
```

```python
haar_features.detectMultiScale(img)
```

```python
# scalerFactor - how much image size is reduced at each step 
faces = haar_features.detectMultiScale(img, scaleFactor=10)
for x,y,w,h in faces:
    # img - on which we want to draw the rectangle
    # (x,y) - top left corner of the rectangle
    # (x+w,y+h) - bottom right corner of the rectangle
    # (0,255,0) - color of the rectangle in BGR format (green in this case)
    # 3 - thickness of the rectangle border
    cv2.rectangle(img,(x,y),(x+w,y+h),(0,255,0),3)

cv2.imshow('output', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
