```python
haar_features = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
```

```python
import cv2

cap = cv2.VideoCapture(0)

while True:
    flag, frame = cap.read()

    if not flag:
        print("Camera not accessible ")
        break

    # scalerFactor - how much image size is reduced at each step 
    # converting to grayscale for better detection
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = haar_features.detectMultiScale(gray, scaleFactor=1.3)

    for x,y,w,h in faces:
        # img - on which we want to draw the rectangle
        # (x,y) - top left corner of the rectangle
        # (x+w,y+h) - bottom right corner of the rectangle
        # (0,255,0) - color of the rectangle in BGR format (green in this case)
        # 3 - thickness of the rectangle border
        cv2.rectangle(frame,(x,y),(x+w,y+h),(0,255,0),3)

    cv2.imshow('output', frame)

    key = cv2.waitKey(25) & 0xFF
    if key == 27: # ESC key to break the loop
        break

# release camera AFTER loop ends
cap.release()
cv2.destroyAllWindows()
```
